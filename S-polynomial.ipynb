```python
def S_polynomial(f, g):
    f = R(f)
    g = R(g)
    LCf = f.leading_coefficient()
    LCg = g.leading_coefficient()
    LMf = f.leading_monomial()
    LMg = g.leading_monomial()
    l = lcm(LMf, LMg)
    
    s = (l/LMf)* 1/LCf * f - (l/LMg) * 1/LCg * g
    return R(s)
```

```python
R.<x,y,z> = PolynomialRing(QQ, order='lex')
f = z^2 + y + z - 1
g = z^3 + y*z + z^2 - z

S = S_polynomial(f, g)
print("S(f,g)=", S)
```
