```python
def divide_by_generators(f, G, R):
    x = R.gen()
    
    p = R(f)
    G = [R(g) for g in G]
    Q = [R(0) for q in G]
    r = R(0)

    while p != 0:
        lc_p = p.leading_coefficient()
        lm_p = p.leading_monomial()
        lmd_p = lm_p.degree()
        
        division_occoured = False
        i = 0
        
        while i < len(G) and not division_occoured:
            
            gi = G[i]

            lc_g = gi.leading_coefficient()
            lm_g = gi.leading_monomial()
            lmd_g = lm_g.degree()
            
            if lm_g.divides(lm_p):
                
                m = (lc_p/lc_g) * x^(lmd_p - lmd_g)
                Q[i] += m
                p -= m * gi

                division_occoured = True
            else:
                i = i + 1
                
        if not division_occoured:
            r += p.leading_term()
            p -= p.leading_term()

    return Q, r
        
```

```python
f  = x^3 - 1
g1 = x^2 - 1
g2 = x - 1

Q, r = divide_by_generators(f, [g1, g2], R)

print("q1 =", Q[0])
print("q2 =", Q[1])
print("r  =", r)
```
