```python
R.<x> = PolynomialRing(QQ)
f = x^3 +6*x^2 - 4*x +7
g = x^2 + 5*x - 3
q = 0
r = f

while r!= 0 and r.degree() >= g.degree():

    lc_r = r.leading_coefficient()
    lm_r = r.leading_monomial()
    lc_g = g.leading_coefficient()
    lm_g = g.leading_monomial()
    
    m = (lc_r / lc_g) * (lm_r / lm_g)
    q += R(m)
    r -= R(m)*g

print("q =", q)
print("r =", r)
```
